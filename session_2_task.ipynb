### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# total salary
# in_hand_salary = total_salary - hra - da - pf - income_task

cost_to_company = float(input("Enter you CTC"))


hra_deduction = 0.1 * cost_to_company
da_deduction = 0.05 * cost_to_company
pf_deduction = 0.03 * cost_to_company
income_tax = 0

if cost_to_company < 500000:
  monthly_in_hand = (cost_to_company - hra_deduction - da_deduction - pf_deduction)/12
  income_tax = 0
  monthly_in_hand = monthly_in_hand - income_tax
elif cost_to_company >= 500000 and cost_to_company < 1000000:
  monthly_in_hand = (cost_to_company - hra_deduction - da_deduction - pf_deduction)/12
  income_tax = 0.1 * (cost_to_company - 500000)
  monthly_in_hand = monthly_in_hand - income_tax
elif cost_to_company >= 1000000 and cost_to_company < 2000000:
  monthly_in_hand = (cost_to_company - hra_deduction - da_deduction - pf_deduction)/12
  income_tax = 0.2 * (cost_to_company - 1000000) + 0.1 * (cost_to_company - 500000)
  monthly_in_hand = monthly_in_hand - income_tax
else:
  monthly_in_hand = (cost_to_company - hra_deduction - da_deduction - pf_deduction)/12
  income_tax = 0.3 * (cost_to_company - 2000000) + 0.2 * (cost_to_company - 1000000) + 0.1 * (cost_to_company - 500000)
  monthly_in_hand = monthly_in_hand - income_tax

print(f"Your in hand monthly salary is {monthly_in_hand}")

````

````python
ctc = float(input("Enter your CTC: "))

hra = 0.10 * ctc
da = 0.05 * ctc
pf = 0.03 * ctc

income_tax = 0
if ctc > 500_000:
    income_tax += 0.10 * (min(ctc, 1_000_000) - 500_000)
if ctc > 1_000_000:
    income_tax += 0.20 * (min(ctc, 2_000_000) - 1_000_000)
if ctc > 2_000_000:
    income_tax += 0.30 * (ctc - 2_000_000)

monthly_in_hand = (ctc - hra - da - pf - income_tax) / 12
print(f"Your in hand monthly salary is {monthly_in_hand}")
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
cost_price=float(input("Enter the cost price of the product"))
selling_price=float(input("Enter the selling price of the product"))

profit_loss = selling_price - cost_price

if profit_loss > 0:
  print(f"The product is in profit of {profit_loss}")
elif profit_loss < 0:
  print(f"The product is in loss of {profit_loss}")
````
