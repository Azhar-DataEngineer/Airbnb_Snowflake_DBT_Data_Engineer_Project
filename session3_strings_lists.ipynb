# Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
cur_population = 10000

for year in range(10,0,-1):
  print(year,cur_population)
  cur_population = cur_population/1.1
```

### Sequence sum

# 1/1! + 2/2! + 3/3! + ...

```python
n= int(input("Enter the number"))
result = 0
fact = 1

for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact

print(result)
```

### Pattern 2

# 1<br>
# 121<br>
# 12321<br>
# 1234321<br>

```python
num = int(input("Enter the number"))

for i in range(1,num+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end="")
  print()
```

# Loop Control Statement

# - Break
# - Continue
# - Pass

```python
  for i in range(1,10):
    if i == 5:
      break
    print(i)
```

```python
print('Demonstrating break in a nested loop:')
for i in range(1, 4):
  print(f'Outer loop: i = {i}')
  for j in range(1, 4):
    if j == 2:
      print(f'  Inner loop: j = {j}, Breaking inner loop!')
      break  # This breaks only the inner loop (for j)
    print(f'  Inner loop: j = {j}')
  print(f'Outer loop continues after inner loop break (if any).')
print('Loop finished.')
```

```python
lower_range = int(input("Enter the lower range"))

higher_range = int(input("Enter the higher range"))


for num in range(lower_range, higher_range):
  for i in range(2,num):
    if num%i == 0:
      break
  else:
    print(num)
```

```python
# continue

for i in range(1,10):
  if i == 5:
    continue
  print(i)
```
